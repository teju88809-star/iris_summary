# ***Descriptive Stastics primer For Iris Data set***



This code loads the Iris dataset, converts it into a Pandas DataFrame, and displays the first five rows.

```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)
df = iris.frame

df.head()
```

**Selecting Columns**
This code selects the sepal length, sepal width, and petal length measurements from the Iris dataset, then displays the first five rows of those columns

```python
columns = [
    "sepal length (cm)",
    "sepal width (cm)",
    "petal length (cm)"
]

data = df[columns]
data.head()
```

**Calculating Summary Statistics**
This code calculates the mean, median, mode, and standard deviation for each selected Iris measurement. It displays the results rounded to two decimal places.





```python
summary = pd.DataFrame({
    "Mean": data.mean(),
    "Median": data.median(),
    "Mode": data.mode().iloc[0],
    "Std Dev": data.std()
})

summary.round(2)
```

**Calculating Percentiles**
This code calculates the 25th, 50th, and 75th percentiles for each measurement and adds them to the summary table. The 25th and 75th percentiles show the range containing the middle 50% of the data.

```python
percentiles = data.quantile([0.25, 0.50, 0.75]).T
percentiles.columns = [
    "25th percentile",
    "50th percentile",
    "75th percentile"
]

summary = summary.join(percentiles)
summary.round(2)
```

Sepal length: Mean and median are close, so the distribution’s center is fairly balanced overall.
Sepal width: Mean and median are also close, indicating little skew in the overall distribution.
Petal length: The mean is below the median because many flowers have short petals, while others have much longer petals. This column also has the largest standard deviation, so its values are most spread out.

Interpretation
The mean and median are close for sepal length and sepal width, suggesting these measurements are fairly balanced overall. Petal length has the largest standard deviation, so its values are more spread out. Its mean is lower than its median, which may indicate skew. The 25th and 75th percentiles show the range containing the middle 50% of values for each measurement.
